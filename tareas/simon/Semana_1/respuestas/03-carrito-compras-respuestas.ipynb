## 🎯 Ahora con tus propios datos

Corre la siguiente celda y escribe productos como `producto,cantidad,precio`. Escribe `fin` para terminar.

````python
carrito_actual = []

print("Ingresa productos como 'producto,cantidad,precio'. Escribe 'fin' para terminar.\n")

while True:
    entrada = input("Producto: ")
    if entrada.strip().lower() == "fin":
        break
    try:
        producto = carrito.parsear_producto(entrada)
        carrito_actual = carrito.agregar_al_carrito(carrito_actual, producto)
        print(f"  ✅ Agregado: {producto['producto']} x{producto['cantidad']}")
    except ValueError as e:
        print(f"  ⚠️ Entrada inválida ({e}). Intenta de nuevo.")

print()
resumen_final = carrito.resumen_carrito(carrito_actual)
print(carrito.formatear_resumen(resumen_final))
````
