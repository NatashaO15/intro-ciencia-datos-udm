El dato estaba en un lado y el comportamiento en otro.

Hoy vas a aprender a empaquetar el dato **y** el comportamiento en un solo lugar: una **clase**. Una clase es un molde para crear objetos; cada objeto (o *instancia*) tiene sus propios datos (**atributos**) y funciones que le pertenecen (**métodos**).


## 1. Tu primera clase: definir e instanciar

- `class NombreClase:` empieza la definición.
- `__init__(self, ...)` es el método especial que se ejecuta al crear el objeto — ahí defines los atributos iniciales.
- `self` es el objeto mismo; así es como un método (o `__init__`) guarda o lee datos del objeto (`self.atributo = valor`).
- Crear un objeto se llama **instanciar**: `mi_objeto = NombreClase(...)`.


```python
class ProductoInventario:
    def __init__(self, nombre, cantidad, precio_unitario):
        self.nombre = nombre.strip().title()
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario


manzanas = ProductoInventario("  manzana", 20, 2500)

print(manzanas.nombre, manzanas.cantidad, manzanas.precio_unitario)
print(type(manzanas))
```

Fíjate: `manzanas` no es un diccionario ni una lista — es un objeto de tipo `ProductoInventario`, con sus propios atributos accesibles con `.` (`manzanas.nombre`, `manzanas.cantidad`, ...). Puedes crear tantos productos como quieras a partir del mismo molde:

```python
peras = ProductoInventario("PERA", 15, 3000)
uvas = ProductoInventario(" uva ", 10, 4500)

for producto in [manzanas, peras, uvas]:
    print(f"{producto.nombre}: {producto.cantidad} unidades a ${producto.precio_unitario:.2f}")
```

## 2. Métodos: funciones que viven dentro del objeto

Un método es una función definida dentro de la clase que recibe `self` como primer parámetro — eso le da acceso a los atributos del objeto sobre el que se llama. Este primer método solo **lee** datos, no cambia nada:

```python
class ProductoInventario:
    def __init__(self, nombre, cantidad, precio_unitario):
        self.nombre = nombre.strip().title()
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario

    def valor_total(self):
        """Valor total del inventario de este producto. No modifica nada."""
        return self.cantidad * self.precio_unitario


peras = ProductoInventario("pera", 15, 3000)
print(f"{peras.nombre}: ${peras.valor_total():.2f} en inventario")
```

## 3. Un método que sí modifica el objeto

### 📖 Ejemplo resuelto — `vender(self, unidades)`

`valor_total` es un método de **solo lectura**: calcula algo pero no cambia el objeto. Este otro, en cambio, **modifica un atributo**.

Compáralo con el trabajo anterior: allá, para cambiar un dato tenías que escribir `registrar_nota(estudiantes, codigo, materia, nota)` — pasarle la lista completa y salir a buscar el diccionario correcto dentro de ella. Aquí el objeto ya sabe cuáles son sus propios datos, así que el método solo tiene que tocar `self`.

Lee este ejemplo con calma: es exactamente el patrón que vas a tener que reproducir en el ejercicio de abajo.

- Si `unidades` es mayor que `self.cantidad`, lanza `ValueError(f"No hay suficiente inventario de {self.nombre} ({self.cantidad} disponibles)")`.
- Si hay suficiente, resta `unidades` de `self.cantidad` con `self.cantidad -= unidades`. El método no devuelve nada (`return`) — su trabajo es *cambiar el estado* del objeto.


```python
class ProductoInventario:
    def __init__(self, nombre, cantidad, precio_unitario):
        self.nombre = nombre.strip().title()
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario

    def valor_total(self):
        return self.cantidad * self.precio_unitario

    def vender(self, unidades):
        if unidades > self.cantidad:
            raise ValueError(f"No hay suficiente inventario de {self.nombre} ({self.cantidad} disponibles)")
        self.cantidad -= unidades


uvas = ProductoInventario("uva", 10, 4500)
print(f"Antes de vender: {uvas.cantidad} unidades")

uvas.vender(4)
print(f"Después de vender 4: {uvas.cantidad} unidades")

try:
    uvas.vender(100)
except ValueError as e:
    print(f"Error esperado: {e}")
```

La clave está en `self.cantidad -= unidades`. A diferencia de una función que recibe un diccionario y devuelve uno nuevo, este método cambia el objeto `uvas` "por dentro": si vuelves a imprimir `uvas.cantidad` en otra celda, seguirá reflejando el descuento. El objeto **recuerda** su nuevo estado.

## 🎯 Cierre: una lista de objetos (mini inventario de tienda)

Así como en el trabajo anterior tenías una lista de diccionarios (`estudiantes`), ahora puedes tener una **lista de objetos**. La diferencia es que cada objeto ya trae sus propios métodos listos para usar.

```python
inventario = [
    ProductoInventario("manzana", 20, 2500),
    ProductoInventario("pera", 15, 3000),
    ProductoInventario("uva", 10, 4500),
    ProductoInventario("kiwi", 25, 1800),
]

valor_total_tienda = sum(producto.valor_total() for producto in inventario)

print(f"Valor total del inventario: ${valor_total_tienda:.2f}")

# Vendamos algunas unidades y veamos cómo cambia el inventario
inventario[0].vender(5)  # 5 manzanas menos
print(f"\nDespués de vender 5 manzanas: {inventario[0].cantidad} unidades, valor total: ${sum(p.valor_total() for p in inventario):.2f}")
```
