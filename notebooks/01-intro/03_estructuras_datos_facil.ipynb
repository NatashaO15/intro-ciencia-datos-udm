## 1. Variables primitivas

Una **variable** es un nombre que usamos para guardar un dato. En Python no hay que avisar de qué tipo es: Python lo descubre solo. Los tipos más básicos son:

- `int`: números enteros (`5`, `-3`)
- `float`: números con decimales (`4.5`)
- `str`: texto, entre comillas (`"Ana"`)
- `bool`: verdadero o falso (`True` / `False`)

Para saber qué tipo tiene algo, usamos `type(...)`.

```python
# Ejemplo: datos de un estudiante
nombre_estudiante = "Ana"
nota = 4.5
aprobado = True

print(nombre_estudiante, type(nombre_estudiante))
print(nota, type(nota))
print(aprobado, type(aprobado))
```

## 2. Strings (texto)

Los strings se pueden "recorrer" por posición (indexación) y "recortar" (slicing). También tienen **métodos** muy útiles para limpiar texto — algo que vamos a necesitar constantemente cuando trabajemos con datos reales (¡en la próxima sesión los vas a usar con Pandas!):

- `.strip()` quita espacios sobrantes al inicio/final.
- `.upper()` / `.lower()` cambian mayúsculas/minúsculas.
- `.title()` pone la Primera Letra De Cada Palabra En Mayúscula.
- `.replace(a, b)` reemplaza texto.

```python
nombre_sucio = "  ana gomez  "
print(repr(nombre_sucio))          # así se ve "crudo", con los espacios visibles

nombre_limpio = nombre_sucio.strip().title()
print(repr(nombre_limpio))

# Indexación y slicing
print(nombre_limpio[0])     # primera letra
print(nombre_limpio[0:3])   # primeras 3 letras
```

## 3. Listas

Una **lista** guarda varios valores en orden, entre corchetes `[ ]`. Podemos:

- Saber cuántos elementos tiene con `len(...)`.
- Sumar todos con `sum(...)`.
- Encontrar el mayor/menor con `max(...)` / `min(...)`.
- Calcular un promedio: `sum(lista) / len(lista)`.

```python
notas = [3.5, 4.0, 2.8, 5.0, 4.2]

cantidad = len(notas)
suma = sum(notas)
promedio = suma / cantidad

print("Cantidad de notas:", cantidad)
print("Suma:", suma)
print("Promedio:", round(promedio, 2))
print("Nota máxima:", max(notas))
print("Nota mínima:", min(notas))
```

## 4. Diccionarios

Un **diccionario** guarda datos con una "etiqueta" (llave) para cada valor, entre llaves `{ }`. En vez de acceder por posición (como en las listas), accedes por nombre.

```python
estudiante = {
    "nombre": "Ana",
    "nota": 4.5,
}

print(estudiante["nombre"])
print(estudiante["nota"])

# f-string: forma cómoda de armar un mensaje con variables adentro
mensaje = f"{estudiante['nombre']} tiene una nota de {estudiante['nota']}"
print(mensaje)
```

## 🎯 Cierre: uniendo todo

Ya tienes las 4 piezas: variables, strings, listas y diccionarios. Ahora vas a recorrer una **lista de diccionarios** — así es como casi siempre se ven los datos reales antes de que existan herramientas como Pandas.

Dado el siguiente grupo de 3 estudiantes, recorre la lista con un `for` e imprime si cada uno está **"Aprobado"** o **"No aprobado"** (nota mayor o igual a 3.0).

```python
grupo = [
    {"nombre": "Ana", "nota": 4.5},
    {"nombre": "Luis", "nota": 2.7},
    {"nombre": "Marta", "nota": 3.0},
]

for estudiante in grupo:
    if estudiante["nota"] >= 3.0:
        estado = "Aprobado"
    else:
        estado = "No aprobado"
    print(f"{estudiante['nombre']}: {estado}")
```
