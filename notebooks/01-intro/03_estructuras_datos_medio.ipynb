## 2. Strings: limpieza y separación

Además de `.strip()`, `.upper()`, `.lower()`, `.title()`, `.replace(a, b)`, vamos a usar `.split(separador)` para partir un string en pedazos — muy útil cuando un dato viene "empaquetado" en una sola línea de texto (como una fila de un CSV sin procesar).

````python
linea_cruda = "manzana,10,2500"

partes = linea_cruda.split(",")
producto, cantidad_str, precio_str = partes

cantidad = int(cantidad_str)
precio = float(precio_str)

print(f"Producto: {producto} | Cantidad: {cantidad} | Precio: {precio:.2f}")
````

### ✏️ Ejercicio 2

Tienes esta lista de nombres de productos, con inconsistencias de formato (espacios y mayúsculas mezcladas):

```python
productos_sucios = ["  manzana", "PERA ", "Uva", "  banano  ", "KIWI"]
```

Recorre la lista con un `for` y arma una nueva lista `productos_limpios` donde cada nombre quede limpio (sin espacios) y en formato título (`Manzana`, `Pera`, etc.).

````python
productos_sucios = ["  manzana", "PERA ", "Uva", "  banano  ", "KIWI"]

# TODO: completa el ciclo
productos_limpios = []
for producto in productos_sucios:
    productos_limpios.append(producto.strip().title())

print(productos_limpios)
````

**Solución** (también se puede escribir como *list comprehension*, que verás más abajo):

````python
productos_sucios = ["  manzana", "PERA ", "Uva", "  banano  ", "KIWI"]
productos_limpios = [p.strip().title() for p in productos_sucios]
print(productos_limpios)
````

## 3. Listas: recorrer, filtrar y contar

Con `for` + `if` podemos recorrer una lista y quedarnos solo con lo que cumple una condición — esto es, en esencia, lo que hace un "filtro" en cualquier herramienta de análisis de datos.

````python
ventas = [23000, 15000, 42000, 8000, 31000, 12000, 19500, 27000, 9000, 50000]

total_dia = sum(ventas)
promedio_dia = total_dia / len(ventas)

ventas_altas = [v for v in ventas if v > 20000]        # list comprehension con condición
contador_altas = len(ventas_altas)

print(f"Total del día: {total_dia}")
print(f"Promedio por venta: {promedio_dia:.2f}")
print(f"Ventas mayores a 20.000: {contador_altas} -> {ventas_altas}")
````

1. Calcula el total y el promedio.
2. Cuenta cuántas ventas fueron menores a 10.000 (posibles "ventas bajas" a revisar).

````python
ventas_sucursal_2 = [18000, 22000, 9500, 34000, 27500, 41000, 5000, 15000, 60000, 21000]

# TODO: completa
total_2 = sum(ventas_sucursal_2)
promedio_2 = total_2 / len(ventas_sucursal_2)
ventas_bajas = [v for v in ventas_sucursal_2 if v < 10000]

print(f"Total: {total_2}")
print(f"Promedio: {promedio_2:.2f}")
print(f"Ventas bajas ({len(ventas_bajas)}): {ventas_bajas}")
````

## 4. Diccionarios: registros de venta

Ahora combinemos todo: una **lista de diccionarios**, donde cada diccionario es un registro (una fila) con varios campos — así se ven casi todos los datos reales antes de cargarlos a una tabla.

````python
registro_ventas = [
    {"producto": "Manzana", "cantidad": 10, "precio_unitario": 2500},
    {"producto": "Pera", "cantidad": 5, "precio_unitario": 3000},
    {"producto": "Uva", "cantidad": 8, "precio_unitario": 4500},
]

total_general = 0
for venta in registro_ventas:
    total_venta = venta["cantidad"] * venta["precio_unitario"]
    total_general += total_venta
    print(f"{venta['producto']}: {venta['cantidad']} unidades -> {total_venta}")

print(f"\nIngreso total del día: {total_general}")
````

## 🎯 Cierre: mini "group by" a mano

Este es exactamente el tipo de tarea que Pandas automatiza con `.groupby()` — pero primero vamos a hacerlo a mano para entender qué hay detrás.

Dado este registro con nombres de producto **sucios** y repetidos, limpia los nombres y luego construye un diccionario `resumen` donde cada llave sea el nombre del producto (limpio) y el valor sea la cantidad total vendida de ese producto (sumando todas sus apariciones).

````python
registro_sucio = [
    {"producto": "  manzana", "cantidad": 10},
    {"producto": "PERA", "cantidad": 5},
    {"producto": "Manzana ", "cantidad": 7},
    {"producto": "pera", "cantidad": 3},
    {"producto": "UVA", "cantidad": 8},
]

resumen = {}
for venta in registro_sucio:
    producto = venta["producto"].strip().title()
    if producto in resumen:
        resumen[producto] += venta["cantidad"]
    else:
        resumen[producto] = venta["cantidad"]

for producto, cantidad_total in resumen.items():
    print(f"{producto}: {cantidad_total} unidades en total")
````

¡Muy bien! Acabas de construir, a mano, el equivalente de un `groupby("producto")["cantidad"].sum()` en Pandas. En la próxima sesión vas a ver cómo esta misma idea se hace en una sola línea sobre tablas completas de datos reales.
